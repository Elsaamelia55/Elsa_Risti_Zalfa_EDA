```python
# TODO D1 — susun tabel penandaan Anda sendiri

# Membuat dictionary PENANDAAN
PENANDAAN = {
    'nama': ('pengenal langsung', 'hapus'),
    'nik': ('pengenal langsung', 'hapus'),
    'email': ('pengenal langsung', 'hash'),
    'no_hp': ('pengenal langsung', 'hash'),
    'alamat': ('pengenal langsung', 'generalisasi'),

    'tanggal_lahir': ('data spesifik', 'generalisasi'),
    'jenis_kelamin': ('data spesifik', 'pertahankan'),
    'pendidikan_terakhir': ('data spesifik', 'generalisasi'),
    'pekerjaan': ('data spesifik', 'generalisasi'),
}

# Membuat DataFrame tanda berdasarkan seluruh kolom dataset
data_tanda = []

for kolom in df.columns:
    if kolom in PENANDAAN:
        kategori, tindakan = PENANDAAN[kolom]
    else:
        kategori, tindakan = 'bukan data pribadi', 'pertahankan'

    data_tanda.append({
        'nama_kolom': kolom,
        'kategori': kategori,
        'tindakan': tindakan
    })

tanda = pd.DataFrame(data_tanda)

# Menampilkan DataFrame tanda
display(tanda)

# Menghitung jumlah kolom per kategori
print("=== JUMLAH KOLOM PER KATEGORI ===")
print(tanda['kategori'].value_counts())

# Memeriksa expected
jumlah_kategori = tanda['kategori'].value_counts()

print("\n=== PEMERIKSAAN EXPECTED ===")
print("Pengenal langsung >= 5:",
      jumlah_kategori.get('pengenal langsung', 0) >= 5)

print("Data spesifik >= 3:",
      jumlah_kategori.get('data spesifik', 0) >= 3)

```

```python
# TODO E1 — HAPUS

# Mengambil nama kolom yang tindakannya adalah hapus
kolom_hapus = tanda.loc[
    tanda['tindakan'] == 'hapus',
    'nama_kolom'
].tolist()

# Membuat DataFrame baru tanpa kolom yang dihapus
aman = df.drop(columns=kolom_hapus)

# Menghitung jumlah kolom yang dibuang dan tersisa
print("Jumlah kolom yang dibuang:", len(kolom_hapus))
print("Jumlah kolom yang tersisa:", aman.shape[1])

# Menampilkan DataFrame aman
display(aman.head())


```

```python
# TODO E2 — HASH

import hashlib
import numpy as np
import pandas as pd

GARAM = 'ganti-dengan-kalimat-rahasia-tim'

# Membuat fungsi samarkan
def samarkan(nilai, garam):
    if pd.isna(nilai) or str(nilai).strip() == '':
        return np.nan

    teks = str(garam) + str(nilai)
    hasil_hash = hashlib.sha256(teks.encode('utf-8')).hexdigest()

    return hasil_hash[:12]


# Mengambil kolom yang memiliki tindakan hash
kolom_hash = tanda.loc[
    tanda['tindakan'] == 'hash',
    'nama_kolom'
].tolist()

# Menerapkan hash pada DataFrame aman
for kolom in kolom_hash:
    if kolom in aman.columns:
        aman[kolom] = aman[kolom].apply(
            lambda x: samarkan(x, GARAM)
        )

# Menampilkan hasil
print("Kolom yang di-hash:", kolom_hash)
display(aman[kolom_hash].head())

# Menguji hasil hash
print("\n=== PENGUJIAN HASH ===")

if kolom_hash:
    kolom_uji = kolom_hash[0]

    nilai_asli = df[kolom_uji].dropna().iloc[0]
    hasil_1 = samarkan(nilai_asli, GARAM)
    hasil_2 = samarkan(nilai_asli, GARAM)

    print("Nilai asli:", nilai_asli)
    print("Hasil hash:", hasil_1)
    print("Nilai asli dan hash berbeda:", str(nilai_asli) != hasil_1)
    print("Nilai sama menghasilkan hash sama:", hasil_1 == hasil_2)

```

```python
# TODO E3 — buktikan hash bersifat konsisten dan tidak terbaca

# a. Hash dua kali nilai yang sama
nilai_1 = samarkan("Zalfa", GARAM)
nilai_2 = samarkan("Zalfa", GARAM)

print("=== PENGUJIAN NILAI SAMA ===")
print("Hash pertama :", nilai_1)
print("Hash kedua   :", nilai_2)
print("Hasil identik:", nilai_1 == nilai_2)


# b. Hash dua nilai berbeda
nilai_3 = samarkan("Zalfa", GARAM)
nilai_4 = samarkan("Anisa", GARAM)

print("\n=== PENGUJIAN NILAI BERBEDA ===")
print("Hash Zalfa :", nilai_3)
print("Hash Anisa :", nilai_4)
print("Hasil berbeda:", nilai_3 != nilai_4)


```

```python
# TODO E4 — GENERALISASI

import pandas as pd
import numpy as np

# Tanggal acuan
ACUAN = pd.Timestamp('2025-09-01')

# a. Menghitung usia peserta berdasarkan tanggal lahir
aman['tanggal_lahir'] = pd.to_datetime(
    aman['tanggal_lahir'],
    errors='coerce'
)

aman['usia'] = (
    ACUAN.year - aman['tanggal_lahir'].dt.year
    - (
        (aman['tanggal_lahir'].dt.month > ACUAN.month) |
        (
            (aman['tanggal_lahir'].dt.month == ACUAN.month) &
            (aman['tanggal_lahir'].dt.day > ACUAN.day)
        )
    ).astype(int)
)

# b. Membuat kelompok umur
batas_umur = [0, 24, 34, 44, 54, float('inf')]
label_umur = ['<25', '25-34', '35-44', '45-54', '55+']

aman['kelompok_umur'] = pd.cut(
    aman['usia'],
    bins=batas_umur,
    labels=label_umur
)

print("=== JUMLAH PESERTA PER KELOMPOK UMUR ===")
display(aman['kelompok_umur'].value_counts().reindex(label_umur))

# Menampilkan kelompok umur terbanyak
print(
    "Kelompok umur terbanyak:",
    aman['kelompok_umur'].value_counts().idxmax()
)

# c. Membuang kolom tanggal_lahir dan kelurahan
aman = aman.drop(
    columns=['tanggal_lahir', 'kelurahan'],
    errors='ignore'
)

print("\nKolom tanggal_lahir dan kelurahan berhasil dibuang.")

# d. Merapikan penulisan nama kecamatan
aman['kecamatan'] = (
    aman['kecamatan']
    .astype('string')
    .str.strip()
    .str.title()
)

print("\n=== HASIL PEMERIKSAAN KECAMATAN ===")
print("Jumlah kecamatan unik:", aman['kecamatan'].nunique())

print("\nDaftar kecamatan:")
print(sorted(aman['kecamatan'].dropna().unique()))

# Pemeriksaan expected
print("\n=== PEMERIKSAAN EXPECTED ===")
print(
    "Kelompok terbanyak 35-44:",
    aman['kelompok_umur'].value_counts().idxmax() == '35-44'
)

print(
    "Kecamatan memiliki 10 nilai unik:",
    aman['kecamatan'].nunique() == 10
)

```

Menghapus nama belum tentu cukup. Untuk setiap kombinasi quasi-identifier, hitung berapa orang yang punya kombinasi sama. Nilai terkecil disebut **k**. Target kelas ini: **k ≥ 5**.

```python
# TODO F4 — TANTANGAN: capai k >= 5
import pandas as pd

# 1. Ambil kolom quasi-identifier dari data aman
kolom_awal = [
    'kecamatan',
    'jenis_kelamin',
    'pendidikan_terakhir',
    'kelompok_umur'
]

data_uji = aman[kolom_awal].copy()

# 2. Fungsi untuk menghitung nilai k
def cek_k(data, kolom):
    ukuran_grup = (
        data.groupby(kolom, dropna=False, observed=True)
        .size()
    )

    k = ukuran_grup.min()
    singleton = (ukuran_grup == 1).sum()

    return k, singleton, len(ukuran_grup)


# 3. Siapkan catatan perubahan
catatan = []

def catat_perubahan(nama, data, kolom):
    k, singleton, jumlah_grup = cek_k(data, kolom)

    catatan.append({
        'Tahap': nama,
        'Kolom quasi-identifier': ', '.join(kolom),
        'Nilai k': k,
        'Jumlah grup': jumlah_grup,
        'Grup berisi 1 orang': singleton
    })

    print(f'{nama}: k = {k}, grup = {jumlah_grup}, singleton = {singleton}')

    return k


# Tahap awal
kolom_aktif = kolom_awal.copy()
k = catat_perubahan('Data awal', data_uji, kolom_aktif)


# 4. Generalisasi kelompok umur
if k < 5:
    data_uji['kelompok_umur'] = (
        data_uji['kelompok_umur']
        .astype('object')
        .replace({
            '<25': '<35',
            '25-34': '<35',
            '35-44': '35-54',
            '45-54': '35-54'
        })
    )

    k = catat_perubahan(
        'Generalisasi kelompok umur',
        data_uji,
        kolom_aktif
    )


# 5. Gabungkan kecamatan dengan jumlah data kurang dari 20
if k < 5:
    frekuensi = data_uji['kecamatan'].value_counts()

    kecamatan_kecil = frekuensi[frekuensi < 20].index

    data_uji['kecamatan'] = data_uji['kecamatan'].replace(
        kecamatan_kecil,
        'Lainnya'
    )

    k = catat_perubahan(
        'Penggabungan kecamatan kecil',
        data_uji,
        kolom_aktif
    )


# 6. Keluarkan pendidikan terakhir dari quasi-identifier
if k < 5:
    kolom_aktif.remove('pendidikan_terakhir')

    k = catat_perubahan(
        'Mengeluarkan pendidikan terakhir',
        data_uji,
        kolom_aktif
    )


# 7. Jika belum mencapai k >= 5, keluarkan kecamatan
if k < 5 and 'kecamatan' in kolom_aktif:
    kolom_aktif.remove('kecamatan')

    k = catat_perubahan(
        'Mengeluarkan kecamatan',
        data_uji,
        kolom_aktif
    )


# 8. Generalisasi kelompok umur lebih jauh
if k < 5 and 'kelompok_umur' in kolom_aktif:
    data_uji['kelompok_umur'] = (
        data_uji['kelompok_umur']
        .astype('object')
        .replace({
            '<35': '<55',
            '35-54': '<55'
        })
    )

    k = catat_perubahan(
        'Generalisasi umur lebih lanjut',
        data_uji,
        kolom_aktif
    )


# 9. Jika masih belum mencapai k >= 5, keluarkan kelompok umur
if k < 5 and 'kelompok_umur' in kolom_aktif:
    kolom_aktif.remove('kelompok_umur')

    k = catat_perubahan(
        'Mengeluarkan kelompok umur',
        data_uji,
        kolom_aktif
    )


# 10. Jika masih belum mencapai k >= 5, keluarkan jenis kelamin
if k < 5 and 'jenis_kelamin' in kolom_aktif:
    kolom_aktif.remove('jenis_kelamin')

    k = catat_perubahan(
        'Mengeluarkan jenis kelamin',
        data_uji,
        kolom_aktif
    )


# 11. Tampilkan riwayat perubahan
riwayat = pd.DataFrame(catatan)

print('\n=== RIWAYAT GENERALISASI ===')
display(riwayat)

print('\n=== HASIL AKHIR ===')
print('Kolom quasi-identifier:', kolom_aktif)
print('Nilai k akhir:', k)

if k >= 5:
    print('BERHASIL: Data telah memenuhi k-anonymity >= 5')
else:
    print('Data masih belum memenuhi k-anonymity >= 5')
```
