## Bagian B — Seleksi dan filter (Latihan 2, berpasangan)
**Tujuan:** mengambil bagian data yang relevan dengan pertanyaan analisis.

````python
# DEMO B1 — memilih kolom

# 1. Memilih satu kolom (Series)
display(df['total_bayar'].head())

# 2. Memilih beberapa kolom (DataFrame)
display(df[['kota', 'kategori', 'total_bayar']].head())
````

````python

# DEMO B2 — loc (berbasis label) vs iloc (berbasis posisi)

# Memastikan DataFrame sudah tersedia
print("Jumlah baris dan kolom:", df.shape)

# 1. Mengambil produk berdasarkan label index 0
print("Produk:", df.loc[0, 'produk'])

# 2. Mengambil data baris pertama kolom ke-5
print("Data kolom ke-5:", df.iloc[0, 4])

# 3. Mengambil index 0 sampai 4 dengan kolom tertentu
display(df.loc[0:4, ['produk', 'jumlah', 'total_bayar']])
````

````python
# DEMO B3 — filter dengan kondisi (boolean mask)
mask = df['total_bayar'] > 1_000_000
print('Transaksi di atas 1 juta:', mask.sum())
df[mask].head()
````

````python
# DEMO B4 — menggabungkan kondisi
hi = df[(df['kategori'] == 'Makanan') & (df['diskon_persen'] >= 10)]
print(hi.shape)
df[df['metode_bayar'].isin(['QRIS','E-Wallet'])].shape
````

````python
# TODO B5

# a. Mengambil transaksi dengan rating pelanggan < 3
df_rating_rendah = df[df['rating_pelanggan'] < 3]

display(df_rating_rendah)

# b. Menghitung jumlah transaksi dan rata-rata total bayar
print("Jumlah transaksi:", len(df_rating_rendah))
print("Rata-rata total bayar:", df_rating_rendah['total_bayar'].mean())

# c. Kategori Kerajinan ATAU Fashion dengan jumlah >= 10
df_kategori = df[
    (df['kategori'].isin(['Kerajinan', 'Fashion'])) &
    (df['jumlah'] >= 10)
]

display(df_kategori)
````

````python

# Transaksi tunai bernilai besar (> 500.000)

df_tunai_besar = df[
    (df['metode_bayar'] == 'Tunai') &
    (df['total_bayar'] > 500000)
]

hasil = df_tunai_besar[
    ['kota', 'produk', 'jumlah', 'total_bayar']
].sort_values(
    by='total_bayar',
    ascending=False
).head(10)

display(hasil)
````

## Bagian C — Kolom turunan dan pengurutan

````python
# DEMO C1 — membuat kolom baru
df['harga_bruto'] = df['jumlah'] * df['harga_satuan']
df['hemat'] = df['harga_bruto'] - df['total_bayar']
df[['harga_bruto','total_bayar','hemat']].head()
````

````python
# DEMO C2 — kolom tanggal
df['tanggal'] = pd.to_datetime(df['tanggal'])
df['bulan'] = df['tanggal'].dt.month
df['nama_hari'] = df['tanggal'].dt.day_name()
df[['tanggal','bulan','nama_hari']].head()
````

````python
# TODO C3

import numpy as np
import pandas as pd

# a. Membuat kolom kelas_nilai
df['kelas_nilai'] = np.where(
    df['total_bayar'] > 500000,
    'Besar',
    'Kecil'
)

# b. Menghitung jumlah transaksi pada setiap kelas
print("Jumlah transaksi berdasarkan kelas nilai:")
print(df['kelas_nilai'].value_counts())

# c. Menampilkan 5 transaksi dengan penghematan (hemat) terbesar
print("\n5 Transaksi dengan Penghematan Terbesar:")

hasil_hemat = df.sort_values(
    by='hemat',
    ascending=False
).head(5)

display(hasil_hemat)

````

## Bagian D — Agregasi dengan groupby
Pola yang akan Anda pakai sepanjang semester: **split → apply → combine**.

````python
# DEMO D1 — agregasi sederhana
df.groupby('kategori')['total_bayar'].mean().round(0)
````

````python
# DEMO D2 — beberapa agregasi sekaligus
ringkas = (df.groupby('kategori')
             .agg(jumlah_transaksi=('id_transaksi','count'),
                  total_penjualan=('total_bayar','sum'),
                  rata_rata_nilai=('total_bayar','mean'),
                  rata_rata_rating=('rating_pelanggan','mean'))
             .round(2)
             .sort_values('total_penjualan', ascending=False))
ringkas
````

````python
# DEMO D3 — dua kunci pengelompokan
df.groupby(['kategori','metode_bayar'])['total_bayar'].sum().unstack(fill_value=0)
````

````python
# TODO D4

# a. Total penjualan per kota

penjualan_kota = df.groupby('kota')['total_bayar'].sum().sort_values(
    ascending=False
)

display(penjualan_kota)
````

````python
# TODO D5 — satu temuan

# Membuat satu kalimat temuan deskriptif

penjualan_kota = df.groupby('kota')['total_bayar'].sum().sort_values(
    ascending=False
)

kota_tertinggi = penjualan_kota.index[0]
nilai_tertinggi = penjualan_kota.iloc[0]

kota_kedua = penjualan_kota.index[1]
nilai_kedua = penjualan_kota.iloc[1]

print(
    f"{kota_tertinggi} memiliki total penjualan sebesar "
    f"Rp{nilai_tertinggi:,.0f}, lebih tinggi dibandingkan "
    f"{kota_kedua} sebesar Rp{nilai_kedua:,.0f}."
)
````
