```python
# TODO E2 — HASH
# Tulis fungsi samarkan(nilai, garam) yang mengembalikan 12 karakter pertama
# dari sha256(garam + nilai). Kembalikan np.nan bila nilainya kosong.
# Terapkan pada kolom yang Anda beri tindakan 'hash'.
# Expected: nilai asli dan hasil hash berbeda, dan nilai asli yang sama menghasilkan hash yang sama

GARAM = 'capybara'
import hashlib
import numpy as np

def samarkan(nilai, garam):
    if pd.isna(nilai):
        return np.nan

    teks = garam + str(nilai)
    hasil_hash = hashlib.sha256(teks.encode()).hexdigest()

    return hasil_hash[:12]
aman['id_peserta'] = aman['id_peserta'].apply(
    lambda nilai: samarkan(nilai, GARAM)
)
aman['id_peserta'] = aman['id_peserta'].apply(
    lambda nilai: samarkan(nilai, GARAM)
)
aman['id_peserta'].head()

```

```python
# TODO E3 — buktikan hash bersifat konsisten dan tidak terbaca
# a. Hash dua kali nilai yang sama, tunjukkan hasilnya identik
nilai = df['id_peserta'].iloc[0]
hash1 = samarkan(nilai, GARAM)
hash2 = samarkan(nilai, GARAM)
print("Hash 1:", hash1)
print("Hash 2:", hash2)
print("Identik:", hash1 == hash2)
# b. Hash dua nilai berbeda, tunjukkan hasilnya berbeda
nilai1 = df['id_peserta'].iloc[0]
nilai2 = df['id_peserta'].iloc[1]
hash1 = samarkan(nilai1, GARAM)
hash2 = samarkan(nilai2, GARAM)
print("Hash nilai 1:", hash1)
print("Hash nilai 2:", hash2)
print("Berbeda:", hash1 != hash2)

```

Jika GARAM ikut tercommit ke GitHub, nilai asli yang sudah di-hash tetap tidak langsung terlihat karena hash tidak dapat dibaca kembali menjadi nilai asli. Namun, perlindungan menjadi lebih lemah jika orang lain mengetahui GARAM dan dapat mencoba membuat hash dari nilai yang mereka tebak untuk mencocokkannya. Jadi, GARAM seharusnya tetap dirahasiakan.

## Bagian F — Memeriksa sisa risiko

Menghapus nama belum tentu cukup. Untuk setiap kombinasi quasi-identifier, hitung berapa orang yang punya kombinasi sama. Nilai terkecil disebut **k**. Target kelas ini: **k ≥ 5**.

```python
# TODO F1 — tulis fungsi cek_k(data, kolom)
# Kembalikan Series berisi: k (grup terkecil), grup berisi 1 orang, total grup
def cek_k(data, kolom):
    grup = data.groupby(kolom).size()
    k = grup.min()
    grup_1_orang = grup[grup == 1]
    total_grup = len(grup)
    return k, grup_1_orang, total_grup

```

```python
# TODO F2 — ukur risiko SEBELUM penanganan
# Gunakan data asli df dengan kolom:
#   ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']
# Expected: k = 1 dan SELURUH 620 baris berada di grup berisi satu orang
df_f2 = df.copy()
df_f2['pendidikan_terakhir'] = df_f2['pendidikan_terakhir'].astype(str)
hasil_sebelum = cek_k(
    df_f2,
    ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']
)
print(hasil_sebelum[0])
print(len(hasil_sebelum[1]))
print(hasil_sebelum[2])
```

**Jawaban Anda:** keempat kolom itu tidak memuat nama maupun nomor telepon. Jelaskan dalam dua kalimat mengapa hasilnya tetap berbahaya.

_Meskipun tidak memuat nama atau nomor telepon, kombinasi kelurahan, jenis kelamin, pendidikan terakhir, dan tanggal lahir dapat menjadi petunjuk yang cukup untuk membedakan seseorang dari peserta lain. Hasil k = 1 menunjukkan bahwa setiap peserta memiliki kombinasi yang unik, sehingga seseorang yang memiliki informasi tersebut berpotensi mengenali peserta tertentu._

```python
# TODO F3 — ukur risiko SESUDAH generalisasi
# Gunakan 'aman' dengan kolom:
#   ['kecamatan', 'jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur']
# Expected: jumlah grup berisi satu orang turun banyak, tetapi k masih 1
aman_f3 = aman.copy()
aman_f3['pendidikan_terakhir'] = aman_f3['pendidikan_terakhir'].astype(str)
aman_f3['kelompok_umur'] = aman_f3['kelompok_umur'].astype(str)
hasil_sesudah = cek_k(
    aman_f3,
    ['kecamatan', 'jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur']
)
print("k =", hasil_sesudah[0])
print("Grup berisi 1 orang =", len(hasil_sesudah[1]))
print("Total grup =", hasil_sudah[2] if False else hasil_sesudah[2])
```

### G3 — Kamus data
Isi `TEMPLATE_Kamus_Data.csv` untuk **setiap** kolom dataset tim, termasuk kolom `kategori_data_pribadi` dan `tindakan_penanganan`. Simpan sebagai `data_dictionary.md` atau `.csv` di repositori.

```python
# G4 — jalankan penanganan pada data tim dan periksa nilai k
# Salin pola Bagian E dan F, ganti dataset dan nama kolomnya.
# Simpan hasilnya ke data/processed/ dan JANGAN unggah data/raw/
import pandas as pd
df = pd.read_csv("/content/diabetes_prediction_dataset.csv")
print("Ukuran dataset:", df.shape)
display(df.head())
```

```python
# G4.1 — Menentukan tindakan penanganan data

tanda = pd.DataFrame({
    'nama_kolom': df.columns,
    'tindakan': [
        'pertahankan',  # gender
        'generalisasi', # age
        'pertahankan',  # hypertension
        'pertahankan',  # heart_disease
        'pertahankan',  # smoking_history
        'pertahankan',  # bmi
        'pertahankan',  # HbA1c_level
        'pertahankan',  # blood_glucose_level
        'pertahankan'   # diabetes
    ]
})

display(tanda)
```

```python
# G4.2 — Membuat DataFrame hasil penanganan

kolom_hapus = tanda.loc[
    tanda['tindakan'] == 'hapus',
    'nama_kolom'
]

aman = df.drop(columns=kolom_hapus).copy()

print("Kolom yang dibuang:", len(kolom_hapus))
print("Kolom yang tersisa:", aman.shape[1])

# Generalisasi usia
aman['kelompok_umur'] = pd.cut(
    aman['age'],
    bins=[-1, 24, 34, 44, 54, float('inf')],
    labels=['<25', '25-34', '35-44', '45-54', '55+'],
    include_lowest=True
)

print("\nDistribusi kelompok umur:")
print(aman['kelompok_umur'].value_counts().sort_index())
```

```python
# G4.3 — Fungsi pemeriksaan k

def cek_k(data, kolom):
    grup = data.groupby(kolom, observed=True).size()
    k = grup.min()
    grup_1_orang = grup[grup == 1]
    total_grup = len(grup)

    return k, grup_1_orang, total_grup

# Sebelum generalisasi
hasil_sebelum = cek_k(
    df,
    ['gender', 'age', 'smoking_history']
)

print("SEBELUM GENERALISASI")
print("k =", hasil_sebelum[0])
print("Grup berisi 1 orang =", len(hasil_sebelum[1]))
print("Total grup =", hasil_sebelum[2])

# Sesudah generalisasi
hasil_sesudah = cek_k(
    aman,
    ['gender', 'kelompok_umur', 'smoking_history']
)

print("\nSESUDAH GENERALISASI")
print("k =", hasil_sesudah[0])
print("Grup berisi 1 orang =", len(hasil_sesudah[1]))
print("Total grup =", hasil_sesudah[2])
print(aman['kelompok_umur'].value_counts().sort_index())
```

```python
import os

os.makedirs('data/processed', exist_ok=True)

aman.to_csv(
    'data/processed/diabetes_aman.csv',
    index=False
)

print("Hasil tersimpan di data/processed/diabetes_aman.csv")
```

```python
grup_satu_orang = hasil_sesudah[1]

display(grup_satu_orang)
```

```python
#G4.4 mencoba meningkatkan nilai k hingga (k=> 5)
#Gabungkan kategori usia
aman['kelompok_umur_baru'] = pd.cut(
    aman['age'],
    bins=[-1, 34, 54, float('inf')],
    labels=['<35', '35-54', '55+'],
    include_lowest=True
)
print(aman['kelompok_umur_baru'].value_counts().sort_index())
#Hitung ulang nilai k
hasil_lebih_luas = cek_k(
    aman,
    ['gender', 'kelompok_umur_baru', 'smoking_history']
)

print("SESUDAH GENERALISASI LEBIH LUAS")
print("k =", hasil_lebih_luas[0])
print("Grup berisi 1 orang =", len(hasil_lebih_luas[1]))
print("Total grup =", hasil_lebih_luas[2])
#Periksa 3 grup tunggal
hasil_lebih_luas[1].reset_index(name='jumlah')
#Periksa jumlah tiap kategori merokok
aman['smoking_history'].value_counts(dropna=False)
#Generalisasi riwayat merokok
aman['riwayat_merokok_baru'] = aman['smoking_history'].replace({
    'ever': 'Riwayat merokok lainnya',
    'former': 'Riwayat merokok lainnya',
    'current': 'Riwayat merokok lainnya',
    'not current': 'Riwayat merokok lainnya'
})
print(aman['riwayat_merokok_baru'].value_counts(dropna=False))
#Hitung k kembali
hasil_merokok = cek_k(
    aman,
    ['gender', 'kelompok_umur_baru', 'riwayat_merokok_baru']
)
print("SESUDAH GENERALISASI RIWAYAT MEROKOK")
print("k =", hasil_merokok[0])
print("Grup berisi 1 orang =", len(hasil_merokok[1]))
print("Total grup =", hasil_merokok[2])
hasil_merokok[1].reset_index(name='jumlah')
aman['gender'].value_counts(dropna=False)
hasil_tanpa_gender = cek_k(
    aman,
    ['kelompok_umur_baru', 'riwayat_merokok_baru']
)

# Menampilkan hasil evaluasi nilai k
print("HASIL PENGUJIAN TANPA GENDER")
print("Nilai k:", hasil_tanpa_gender[0])
print("Jumlah grup berisi 1 orang:", len(hasil_tanpa_gender[1]))
print("Jumlah seluruh grup:", hasil_tanpa_gender[2])
```
