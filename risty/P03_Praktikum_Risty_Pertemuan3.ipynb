## Bagian E — Menerapkan tiga tindakan

```python
# TODO E1 — HAPUS

# Ambil nama kolom yang tindakannya 'hapus'
kolom_hapus = tanda.loc[tanda['tindakan'] == 'hapus', 'nama_kolom'].tolist()

# Buat DataFrame baru tanpa kolom yang harus dihapus
aman = df.drop(columns=kolom_hapus)

# Cetak jumlah kolom yang dibuang dan tersisa
print("Jumlah kolom yang dibuang:", len(kolom_hapus))
print("Jumlah kolom yang tersisa:", aman.shape[1])
```

```python
# TODO E2 — HASH

import hashlib
import numpy as np

GARAM = 'ganti-dengan-kalimat-rahasia-tim'

def samarkan(nilai, garam):
    if pd.isna(nilai) or nilai == '':
        return np.nan

    teks = str(nilai)
    hasil = hashlib.sha256((garam + teks).encode()).hexdigest()
    return hasil[:12]

# Ambil kolom yang diberi tindakan 'hash'
kolom_hash = tanda.loc[tanda['tindakan'] == 'hash', 'nama_kolom'].tolist()

# Terapkan hash
for kolom in kolom_hash:
    aman[kolom] = aman[kolom].apply(
        lambda x: samarkan(x, GARAM)
    )

print("Kolom yang di-hash:", kolom_hash)
```

```python
# TODO E3 — buktikan hash bersifat konsisten dan tidak terbaca

# a. Hash dua kali nilai yang sama
nilai1 = "Risty"
hash1 = samarkan(nilai1, GARAM)
hash2 = samarkan(nilai1, GARAM)

print("Hash nilai yang sama:")
print("Hash 1:", hash1)
print("Hash 2:", hash2)
print("Apakah identik?", hash1 == hash2)

# b. Hash dua nilai berbeda
nilai2 = "Risty"
nilai3 = "Rina"

hash3 = samarkan(nilai2, GARAM)
hash4 = samarkan(nilai3, GARAM)

print("\nHash nilai berbeda:")
print("Hash Risty:", hash3)
print("Hash Rina :", hash4)
print("Apakah berbeda?", hash3 != hash4)
```

**Jawaban Anda:** kalau `GARAM` ikut ter-*commit* ke GitHub, apa yang masih terlindungi dan apa yang tidak?

_(tulis di sini)_

```python
# TODO E4 — GENERALISASI

ACUAN = pd.Timestamp('2025-09-01')

# a. Hitung usia peserta
aman['usia'] = ACUAN.year - aman['tanggal_lahir'].dt.year

# Kurangi 1 jika ulang tahun belum terjadi pada tanggal acuan
belum_ulang_tahun = (
    (aman['tanggal_lahir'].dt.month > ACUAN.month) |
    (
        (aman['tanggal_lahir'].dt.month == ACUAN.month) &
        (aman['tanggal_lahir'].dt.day > ACUAN.day)
    )
)

aman.loc[belum_ulang_tahun, 'usia'] -= 1

# b. Buat kelompok umur
aman['kelompok_umur'] = pd.cut(
    aman['usia'],
    bins=[-1, 24, 34, 44, 54, float('inf')],
    labels=['<25', '25-34', '35-44', '45-54', '55+']
)

print("Jumlah peserta per kelompok umur:")
print(aman['kelompok_umur'].value_counts().sort_index())

# c. Buang kolom tanggal_lahir dan kelurahan
aman = aman.drop(columns=['tanggal_lahir', 'kelurahan'])

# d. Rapikan penulisan kecamatan
aman['kecamatan'] = aman['kecamatan'].str.strip().str.title()

print("\nJumlah nilai unik kecamatan:", aman['kecamatan'].nunique())
```

## Bagian F — Memeriksa sisa risiko

Menghapus nama belum tentu cukup. Untuk setiap kombinasi quasi-identifier, hitung berapa orang yang punya kombinasi sama. Nilai terkecil disebut **k**. Target kelas ini: **k ≥ 5**.

```python
# TODO F1 — tulis fungsi cek_k(data, kolom)

def cek_k(data, kolom):
    ukuran_grup = data.groupby(kolom, dropna=False).size()

    k = ukuran_grup.min()
    grup_satu_orang = (ukuran_grup == 1).sum()
    total_grup = ukuran_grup.size

    return pd.Series({
        'k': k,
        'grup_berisi_1_orang': grup_satu_orang,
        'total_grup': total_grup
    })

```

```python
# TODO F2 — ukur risiko SEBELUM penanganan

kolom_asli = [
    'kelurahan',
    'jenis_kelamin',
    'pendidikan_terakhir',
    'tanggal_lahir'
]

hasil_sebelum = cek_k(df, kolom_asli)

print(hasil_sebelum)
```

**Jawaban Anda:** keempat kolom itu tidak memuat nama maupun nomor telepon. Jelaskan dalam dua kalimat mengapa hasilnya tetap berbahaya.

_(tulis di sini)_

```python
# TODO F3 — ukur risiko SESUDAH generalisasi

kolom_aman = [
    'kecamatan',
    'jenis_kelamin',
    'pendidikan_terakhir',
    'kelompok_umur'
]

hasil_sesudah = cek_k(aman, kolom_aman)

print(hasil_sesudah)
```

```python
# TODO F4 — TANTANGAN: capai k >= 5

# Pilihan 1: keluarkan pendidikan_terakhir dari quasi-identifier
kolom_f4 = [
    'kecamatan',
    'jenis_kelamin',
    'kelompok_umur'
]

hasil_f4 = cek_k(aman, kolom_f4)

print("Hasil setelah generalisasi:")
print(hasil_f4)

print("\nKolom yang digunakan:", kolom_f4)
print("Dampak: pendidikan_terakhir dikeluarkan dari quasi-identifier untuk mengurangi")
print("jumlah grup kecil dan meningkatkan nilai k.")

```

```python
# G4 — jalankan penanganan pada data tim dan periksa nilai k

# 1. Salin data tim
aman_tim = df.copy()

# 2. Hapus kolom yang ditandai "hapus"
kolom_hapus_tim = tanda.loc[
    tanda['tindakan'] == 'hapus', 'nama_kolom'
].tolist()

aman_tim = aman_tim.drop(columns=kolom_hapus_tim)

# 3. Hash kolom yang ditandai "hash"
GARAM = 'ganti-dengan-kalimat-rahasia-tim'

for kolom in tanda.loc[tanda['tindakan'] == 'hash', 'nama_kolom']:
    aman_tim[kolom] = aman_tim[kolom].apply(
        lambda x: samarkan(x, GARAM)
    )

# 4. Generalisasi tanggal lahir menjadi kelompok umur
ACUAN = pd.Timestamp('2025-09-01')

aman_tim['usia'] = ACUAN.year - aman_tim['tanggal_lahir'].dt.year

belum_ulang_tahun = (
    (aman_tim['tanggal_lahir'].dt.month > ACUAN.month) |
    (
        (aman_tim['tanggal_lahir'].dt.month == ACUAN.month) &
        (aman_tim['tanggal_lahir'].dt.day > ACUAN.day)
    )
)

aman_tim.loc[belum_ulang_tahun, 'usia'] -= 1

aman_tim['kelompok_umur'] = pd.cut(
    aman_tim['usia'],
    bins=[-1, 24, 34, 44, 54, float('inf')],
    labels=['<25', '25-34', '35-44', '45-54', '55+']
)

# 5. Hapus data yang terlalu spesifik
aman_tim = aman_tim.drop(columns=['tanggal_lahir', 'kelurahan'])

# 6. Rapikan kecamatan
aman_tim['kecamatan'] = aman_tim['kecamatan'].str.strip().str.title()

# 7. Periksa nilai k sebelum penanganan
kolom_sebelum = [
    'kelurahan',
    'jenis_kelamin',
    'pendidikan_terakhir',
    'tanggal_lahir'
]

print("Nilai k sebelum penanganan:")
print(cek_k(df, kolom_sebelum))

# 8. Periksa nilai k setelah penanganan
kolom_sesudah = [
    'kecamatan',
    'jenis_kelamin',
    'pendidikan_terakhir',
    'kelompok_umur'
]

print("\nNilai k setelah penanganan:")
print(cek_k(aman_tim, kolom_sesudah))

# 9. Simpan hasil ke data/processed/
import os

os.makedirs('data/processed', exist_ok=True)

aman_tim.to_csv(
    'data/processed/data_tim_aman.csv',
    index=False
)

print("\nData berhasil disimpan ke data/processed/data_tim_aman.csv")
```
