```python
import pandas as pd
import numpy as np

df = pd.read_csv('kunjungan_perpustakaan_2025.csv')
df.head()
```

### Tugas 2 (20 poin) — Filter

```python
# a. Cetak jumlah baris dan kolom
print("a. Jumlah baris dan kolom:")
print(df.shape)

# b. Cetak jumlah nilai kosong (missing) untuk setiap kolom
print("\nb. Jumlah nilai kosong setiap kolom:")
print(df.isnull().sum())

# c. Cetak berapa mahasiswa unik (nim_hash) yang pernah berkunjung
print("\nc. Jumlah mahasiswa unik yang pernah berkunjung:")
print(df['nim_hash'].nunique())
```

---
### Tugas 3 (20 poin) — Kolom turunan

```python
# Ambil kunjungan mahasiswa prodi Teknik Informatika
# dengan durasi_menit > 120
data_ti = df[
    (df['prodi'] == 'Teknik Informatika') &
    (df['durasi_menit'] > 120)
]

# Cetak jumlah baris
print("Jumlah baris:", len(data_ti))

# Cetak rata-rata jumlah_pinjam (2 desimal)
print("Rata-rata jumlah_pinjam:", round(data_ti['jumlah_pinjam'].mean(), 2))
```

---
### Tugas 4 (25 poin) — Agregasi

```python
# Buat tabel ringkasan per jenis_ruang
ringkas = df.groupby('jenis_ruang').agg(
    jumlah_kunjungan=('jenis_ruang', 'size'),
    rata_rata_durasi=('durasi_menit', 'mean'),
    total_jumlah_pinjam=('jumlah_pinjam', 'sum'),
    rata_rata_kepuasan=('kepuasan', 'mean')
).reset_index()

# Pembulatan
ringkas['rata_rata_durasi'] = ringkas['rata_rata_durasi'].round(1)
ringkas['rata_rata_kepuasan'] = ringkas['rata_rata_kepuasan'].round(2)

# Urutkan dari rata-rata durasi terbesar
ringkas = ringkas.sort_values(
    by='rata_rata_durasi',
    ascending=False
)

# Tampilkan
display(ringkas)
```

---
### Tugas 5 (10 poin) — Menyimpan hasil

```python
# Simpan tabel 'ringkas' ke file CSV
ringkas.to_csv('ringkasan_ruang.csv', index=False)

print("File berhasil disimpan sebagai 'ringkasan_ruang.csv'")
```
